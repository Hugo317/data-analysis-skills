## 6. Data quality checks

```python
print('Duplicate rows:', df.duplicated().sum())
print('Duplicate User IDs:', df['User ID'].duplicated().sum())
```

```python
for col in df.select_dtypes(include='object').columns:
    print(f'--- {col} ---')
    print(df[col].value_counts(dropna=False))
    print()
```

**Flagged issue: `Location` is fragmented by inconsistent formatting.**

- `Central Park,NY` (24) vs `Central Park,ny` (8) — casing inconsistency, same location.
- `Market City, NY` (20) vs `Market City, MY` (2) — looks like a data-entry typo (NY → MY), not a real second city.
- `China Town, NY` has a leading space after the comma that other entries don't.

10 distinct values likely represent ~8 real locations. Any location-level ranking below is provisional until this is cleaned.

## 7. Numerical distributions and relationships

```python
# Check whether Overall Rating is an independent measurement or derived
calc_overall = (df['Food Rating'] + df['Service Rating']) / 2
print('Overall Rating == mean(Food, Service) for all rows?', (calc_overall - df['Overall Rating']).abs().lt(1e-9).all())
```

`Overall Rating` is exactly `(Food Rating + Service Rating) / 2` for every row — it is derived, not an independent signal. Treat Food and Service as the two real outcome variables.

```python
df['Age'] = 2026 - df['YOB']
df[['Budget', 'Food Rating', 'Service Rating', 'Overall Rating', 'Age']].corr(numeric_only=True).round(2)
```

Budget and Age show essentially no correlation with any rating (|r| < 0.15). Food Rating and Service Rating are only weakly correlated with each other (0.08) — a diner's food experience and service experience move largely independently.

```python
df['Overall Rating'].value_counts().sort_index()
```

## 8. Business context: Cuisines as "product", Location as "region"

This is a restaurant-dining survey, not a revenue/profit dataset. There is no company, revenue, or cost data. `Overall Rating` is treated as the business KPI in place of profit; `Cuisines` as the product dimension; `Location` as the region dimension.

## 9. Cuisine performance

```python
cuisine_perf = df.groupby('Cuisines').agg(
    n=('User ID', 'count'),
    avg_overall=('Overall Rating', 'mean'),
    avg_food=('Food Rating', 'mean'),
    avg_service=('Service Rating', 'mean'),
).sort_values('avg_overall', ascending=False).round(2)
cuisine_perf
```

Japanese rates highest overall (3.60); Filipino lowest (3.04). Notable: **Indian and Filipino have the two lowest Food scores but above-average Service scores** — food and service diverge in opposite directions for these cuisines, which the single blended `Overall Rating` masks.

## 10. Location performance (provisional — raw, uncleaned)

```python
location_perf = df.groupby('Location').agg(
    n=('User ID', 'count'),
    avg_overall=('Overall Rating', 'mean'),
).sort_values('avg_overall', ascending=False).round(2)
location_perf
```

`St. George,NY` is the largest location by volume (46 records) but rates below average (3.12). This table will shift once the casing/typo issues from the data-quality section are fixed.

## 11. Segment analysis

```python
for col in ['Marital Status', 'Activity', 'Gender', 'Alcohol', 'Smoker', 'Often A S']:
    print(f'--- {col} ---')
    print(df.groupby(col)['Overall Rating'].agg(['count', 'mean']).round(2))
    print()
```

- **Marital Status**: Divorced diners rate much higher (4.54, n=14) than Married (2.87, n=86) or Single (3.34, n=100). Direction is clear but the Divorced group is small — worth confirming it holds up, not a settled finding.
- **Smoker**: "Often" smokers average 3.89 vs "Never" 2.74 and "Socially" 2.98 — a large, non-monotonic gap with no obvious causal story. Flagged as curious, not explained.
- **Alcohol**: cleaner gradient — Never (2.97) < Often (3.39) < Socially (3.47).
- **`Often A S`**: column name is ambiguous (possibly "Often Asks for Substitutions/Special"); Yes rates much higher (4.10) than No (3.09). Confirm the column's actual meaning before relying on it.

```python
df.groupby('Budget')['Overall Rating'].agg(['count', 'mean']).round(2)
```

Budget shows no clear pattern (flat 3.14–3.27 across tiers 3–5); the Budget=1 bucket looks higher (3.85) but n=10, likely noise.

## Assumptions

- Age computed as `2026 − YOB` (no explicit "today" reference in the data).
- Treated `Cuisines` as the product dimension and `Location` as the region dimension, since there is no revenue/profit/company data.
- Location casing/typo issues were **not** normalized before computing the location table — those numbers are provisional.

## Questions this dataset can answer next

## Final summary

**FIRST ANALYSIS SUMMARY**

Dataset: 200 rows × 15 columns, no missing values, no duplicates.

Data quality: ⚠️ `Location` fragmented by casing/typo (10 values → ~8 real locations). ⚠️ `Overall Rating` is derived (mean of Food + Service), not independent. ⚠️ `Often A S` has an ambiguous, unconfirmed meaning.

Business (rating as KPI): mean Overall Rating 3.23 (std ~1.08), spread fairly evenly from 1 to 5.

Cuisines: Japanese rates highest (3.60); Filipino lowest (3.04). Indian and Filipino show an unusual food-low / service-high split.

Locations: `Market City, NY` and `Central Park` lead (provisional); `St. George,NY`, the largest location by volume, rates below average.

Most interesting findings:
- Food and Service ratings are largely independent of each other (r=0.08).
- Divorced diners and "Often" smokers rate notably higher than other segments — real patterns in the data, not explained causally, and based on modest sample sizes.
- Budget and Age have no meaningful relationship with satisfaction.

Recommended next analysis:
1. Clean `Location` and re-run the location ranking.
2. Investigate the Food/Service divergence by cuisine, controlling for location.
3. Confirm the meaning of `Often A S` and re-test its association with rating.
