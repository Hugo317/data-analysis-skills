# Analysis Summary

Running index of every insight investigated against `Cuisine_rating_clean.csv`. Each row links to its dedicated notebook, which is the full analysis and the source of truth — this page is a lightweight index, updated after every completed insight.

## Insight 1 — Does smoking affect the way people enjoy food?

**Notebook:** [`01_smoking_food_enjoyment.ipynb`](./01_smoking_food_enjoyment.ipynb)

**Headline conclusion:** Yes, but not gradually. `Never` (mean Food Rating 2.83) and `Socially` (2.68) smokers don't differ from each other, while `Often` smokers rate food dramatically higher (4.10) — a statistically significant (Kruskal-Wallis p<0.0001) and practically large (~1.3-1.4 point) jump concentrated entirely in the heaviest-smoking group. Service Rating shows a milder version of the same pattern. This is association only, observational data — the step-function shape (rather than a dose-response gradient) is flagged as more consistent with a confound than a direct smoking effect; a follow-up cross-tab against Marital Status/Cuisines/Location was proposed to test that.

## Insight — What types of cuisines are most popular by location and by gender?

**Notebook:** [`03_cuisine_popularity_by_location_gender.ipynb`](./03_cuisine_popularity_by_location_gender.ipynb)

**Headline conclusion:** By location, the top cuisine varies place to place and no location has a dominant favorite — leads are narrow (mostly 22–33% of a location's diners), with China Town,NY's lean toward Seafood (36.4%) the strongest single result; Market City,NY is a genuine tie between Indian and Japanese. By gender, Female diners lean most toward French (19.5%), while Male diners have no single leader — Filipino, Indian, and Japanese are exactly tied at 22 diners each (18.6%). Italian is the least popular cuisine overall, especially among Male diners (5.1%). Popularity here is demand/frequency-based, not rating-based — a chi-square test and a popularity-vs-satisfaction comparison were flagged as natural follow-ups.

## Insight — What kinds of age enjoy food and service more?

**Notebook:** [`02_age_food_service_enjoyment.ipynb`](./02_age_food_service_enjoyment.ipynb)

**Headline conclusion:** Age affects both, but non-monotonically, and food and service peak in different groups — the raw age-rating correlation is ~0 (a naive check would miss this). Once bracketed, the effect is significant (Food: ANOVA p=0.001; Service: p=0.023). The youngest diners (≤24) enjoy food most (mean 3.72); the 45-54 bracket gets the best service (mean 3.92); diners in their late 30s/early 40s (35-44) are the least satisfied on both fronts (food 2.46, service 2.79). The 35-44 dip may be confounded with Marital Status (that bracket is 57% Married, and Married diners rate lower generally) rather than a pure age effect — flagged as a follow-up.

## Insight — Does YOB relate to Budget and to Area?

**Notebook:** [`04_yob_budget_area.ipynb`](./04_yob_budget_area.ipynb)

**Headline conclusion:** Budget — yes, but non-linearly: raw age-budget correlation is ~0, yet the age-bracket ANOVA is significant (p=0.0037) because 25-34-year-olds budget noticeably lower (mean 3.36/5) than every other bracket (which cluster 3.6-4.3) — a dip, not a trend, so a naive correlation check misses it (same shape as the age/food-service insight above). Area — only a mild, unconfirmed lead: `Central Park,NY` skews youngest (mean age 35.7) and `Market City,NY` skews oldest (48.7), but the cross-location difference doesn't clear significance (p≈0.07). Also surfaced: `Area code` is *not* redundant with `Location` as assumed going in — it's a much finer-grained code (up to 22 distinct codes within one location) — flagged as a follow-up worth its own insight.

## Insight — Check marital status per area

**Notebook:** [`05_marital_status_by_area.ipynb`](./05_marital_status_by_area.ipynb)

**Headline conclusion:** Marital status clusters by location rather than spreading evenly. `Central Park,NY` skews heavily Single (81.2% vs. 50.0% dataset-wide, n=32) and `Market City,NY` skews heavily Married (72.7% vs. 43.0% overall, n=22) — the two best-supported patterns. A chi-square test across all 8 locations is significant (p=0.0037) and holds after dropping the smallest location, but 42% of the contingency table's cells have an expected count under 5, so the omnibus test is directional evidence, not proof — the two standout locations carry the finding. This independently corroborates `04_yob_budget_area.ipynb`'s finding that Central Park skews youngest and Market City skews oldest by age. Also flagged as relevant to the smoking/food-rating confound question from Insight 1 (marital-status mix varying by location is a plausible indirect channel, not yet directly tested).
